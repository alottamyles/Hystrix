# Step 1: Classify Customer Issue

You are a customer support assistant for an online retailer. Analyze the customer's message and classify the issue into exactly one of these categories:

- Missing Delivery
- Damaged Item
- Refund Request
- Wrong Item Sent
- Account Issue
- Other

Return only:
Category: [category]
Reason: [one brief sentence]

Customer message:
"Hi, my package says it was delivered yesterday, but I never received it. I checked around my front door and asked my neighbors, but nobody has seen it. What should I do?"

```python
customer_message = """
Hi, my package says it was delivered yesterday, but I never received it.
I checked around my front door and asked my neighbors, but nobody has seen it.
What should I do?
"""

step1_output = """
Category: Missing Delivery
Reason: The package is marked as delivered, but the customer cannot locate it.
"""

print(step1_output)
```

# Step 2: Identify Missing Info

You are a customer support assistant for an online retailer. The previous step classified the customer's issue as:

[Step 1 output]

Based on the classification of the issue and the customer's original message, find any additional info needed before resolving the issue.

Constraints:
- Ask for no more than 3 pieces of information
- Do not ask for sensitive info such as passwords or credit/debit card numbers
- Do not repeat information the customer has already provided
- Speak politely and with concise tone

Return a numbered list only.

```python
step2_prompt = f"""
You are a customer support assistant for an online retailer.

The previous step classified the customer's issue as:

{step1_output}

Based on that classification and the customer's original message, determine what additional information is needed before resolving the issue.

Customer message:
{customer_message}

Constraints:
- Ask for no more than 3 pieces of information.
- Do not ask for sensitive information such as passwords or credit/debit card numbers.
- Do not repeat information the customer has already provided.
- Use a polite and concise tone.

Return a numbered list only.
"""

print(step2_prompt)
```

```python
step2_output = """
1. Please provide your order number.
2. Please confirm the delivery address on the order.
3. Please let us know whether you checked the tracking details or delivery photo from the carrier.
"""

print(step2_output)
```

# Step 3: Propose Solution

You are a customer support assistant for an online retailer.

The customer's issue was classified as:

[Step 1 output]

The following additional info was identified as necessary:

[Step 2 output]

Assume the customer has confirmed:
- The order number is valid
- The delivery address is correct
- The package still cannot be located

Recommend the next appropriate step

Constraints:
- Be professional and helpful
- Do not guarantee a refund or replacement
- Give the customer clear next steps
- Keep the response concise

Return:

Recommended Action:

Customer Response:

```python
step3_prompt = f"""
You are a customer support assistant for an online retailer.

The customer's issue was classified as:

{step1_output}

The following additional info was identified as necessary:

{step2_output}

Assume the customer has now confirmed:
- The order number is valid
- The delivery address is correct
- The package still cannot be located

Recommend the next appropriate step.

Constraints:
- Be professional and helpful.
- Do not guarantee a refund or replacement.
- Give the customer clear next steps.
- Keep the response concise.

Return:

Recommended Action:

Customer Response:
"""

print(step3_prompt)
```

```python
step3_output ="""
Recommended Action:
Begin an investigation into the missing package and verify the shipment with the carrier.

Customer Response:
Thank you for confirming those details. Since the delivery address is correct and the package still cannot be located, the next step is to review the tracking information and begin an investigation. If necessary, the case can be reviewed for a possible replacement, refund, or shipment investigation.

"""

print(step3_output)
```

# Step 4: Decide Whether to Escalate

You are a customer support assistant for an online retailer.

The customer's issue was classified as:

[Step 1 output]

The additional information identified was:

[Step 2 output]

The recommended action was:

[Step 3 output]

Decide whether the case should be escalated to a human support agent.

Escalate if..
1. Authorization is required for refund or replacement
2. Issue cannot be resolved through normal troubleshooting
3. Carrier investigation may be required.

Return only:

Escalation Required: Yes or No
Reason: [one brief sentence]


```python
step4_prompt = f"""
You are a customer support assistant for an online retailer.

The customer's issue was classified as:

{step1_output}

The additional information identified was:

{step2_output}

The recommended action was:

{step3_output}

Decide whether the case should be escalated to a human support agent.

Escalate if:
- A refund or replacement may require authorization.
- The issue cannot be resolved through normal troubleshooting.
- A carrier investigation may be required.

Return only:

Escalation Required: Yes or No
Reason: [one brief sentence]
"""

print(step4_prompt)
```

```python
step4_output = """
Escalation Required: Yes
Reason: The package is still missing after basic troubleshooting and may require authorization or a carrier investigation.
"""

print(step4_output)
```
