# Plan / Reason

The function will add all numbers in the list, divide the total by the number of values.

The program will initially be tested with a normal list.

After reviewing the output, the empty-list case will be tested as well to determine whether the code needs to be improved or not.

```python
def calculate_average(numbers):
  return sum(numbers) / len(numbers)

numbers = [10, 20, 30, 40, 50]
result = calculate_average(numbers)
print(result)
```

# Observe

Function works correctly for the sample list and returns an average of 30.0.

However, it can also failt if the input list is empty because it attempts to divide by the number of values in the list.

```python
empty_numbers = ([])

def calculate_average(empty_numbers):
  return sum(empty_numbers) / len(empty_numbers)
result = calculate_average(empty_numbers)

print(result)
```

# Fix

The empty-list test produced a ZeroDivisionError because the function tried to divide by zero.

How do we fix this? The function has to first check whether the list is empty. If it is, the function will return a message instead of calculating.

```python
def calculate_average(numbers):
  if len(numbers) == 0:
    return "Cannot calculate average of an empty list."

  return sum(numbers) / len(numbers)

numbers = [10, 20, 30, 40, 50]
result = calculate_average(numbers)
print(result)

empty_numbers = []
empty_result = calculate_average(empty_numbers)
print(empty_result)
```
