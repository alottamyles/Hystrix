## Original Prompt
Summarize the benefits of regular exercise in one short, concise paragraph for a general audience.

## Original Summary
Regular exercise leads to improved physical health, increased energy, support for mental well-being, and helps maintain a healthy weight. It can also strengthen muscles and improve cardiovascular health.

## Self-Reflection Prompt
Review the summary above and evaluate it using the following criteria;
- Are the statements reasonable and accurate?
- Is the summary easy to understand?
- Is the language appropriate for a general audience?
- Keep word count below 80 words
- Focus only on the main benefits of regular exercise
- Use one concise paragrah

Be sure to identify weaknesses in the original summary, then create a revised version that addresses them.

```python
original_summary = """
Regular exercise leads to improved physical health, increased energy, support for mental well-being, and helps maintain a healthy weight. It can also strengthen muscles and improve cardiovascular health.
"""

print("Original Summary:")
print(original_summary)
```

## Self-Critique:
While the original summary is clear and accurate, it is somewhat general and could be improved by specifying more on the benefits while still keeping the language simple and concise for a general audience.

## Improved Summary:
Regular exercise can help improve cardiovascular health, strengthens muscles, manage weight, and increase energy levels. It also supports mental health by reducing stress and improving mood. Overall, consistent physical activity is key to improving pboth physical health and quality of life.

```python
improved_summary = """
Regular exercise can help improve cardiovascular health, strengthens muscles, manage weight, and increase energy levels. It also supports mental health by reducing stress and improving mood. Overall, consistent physical activity is key to improving both physical health and quality of life.
"""

print("Improved Summary:")
print(improved_summary)

word_count = len(improved_summary.split())
print("\nImproved Summary Word Count:", word_count)
```
